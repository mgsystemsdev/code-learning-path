# 📁 Functions


### Functions in Python are used to organize reusable blocks of code.  
A function is a named sequence of statements that perform a specific task. Functions help break programs into smaller, modular pieces, making code easier to read, maintain, and reuse.

Functions can take input (parameters), perform operations, and return output (values). They are one of the most important building blocks in Python programming.

---

**Key features of functions:**
- Defined using the `def` keyword followed by the function name and parentheses.
- Can take zero or more parameters (inputs).
- Can return a value using the `return` statement.
- Improve code reusability and readability.
- Support default parameters, keyword arguments, and variable-length arguments (`*args`, `**kwargs`).
- Python also supports anonymous functions using the `lambda` keyword.

---

**Parts of a function:**
- **Function name**: A unique identifier that describes what the function does. Used to call (invoke) the function later.
- **Parameters**: Variables listed inside parentheses after the function name. They act as placeholders for values (arguments) passed into the function.
- **Body**: The indented block of code inside the function. This is where the logic or task of the function is written.
- **Return value**: The output of the function, specified with the `return` keyword. If no return is provided, the function returns `None`.

---

**Common operations:**
- Defining a function with `def`.
- Calling a function by its name followed by parentheses.
- Passing arguments to customize behavior.
- Returning values for later use.
- Using `*args` for arbitrary positional arguments.
- Using `**kwargs` for arbitrary keyword arguments.

---

**Use cases of functions:**
- Encapsulating repetitive tasks into reusable code.
- Organizing large programs into smaller, logical sections.
- Reducing redundancy and improving maintainability.
- Implementing modular design and separation of concerns.
- Supporting functional programming with higher-order functions.

---

👉 Functions are essential for structuring Python programs, allowing you to write cleaner, more efficient, and more maintainable code.



# 📁 Defining a function


### Simple Function


```python
def greet_user():
    """
    This function greets the user.
    """
    print("Hello, world!")

greet_user()
```

### Passing information to a function

```python
def greet_user(username):
    """Display a simple greeting."""
    print(f"Hello, {username.title()}!")

greet_user('jesse')
```

# 📁 Positional Argument

### Posional arguments


```python
def describe_pet(animal_type, pet_name):
    """Display information about a pet."""
    print(f"\nI have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")

describe_pet('hamster', 'harry')
```

### Multiple functions calls


```python
def describe_pet(animal_type, pet_name):
    """Display information about a pet."""
    print(f"\nI have a {animal_type}.")
    print(f"My {animal_type}'s name is {pet_name.title()}.")

describe_pet('hamster', 'harry')
describe_pet('dog', 'willie')
```
