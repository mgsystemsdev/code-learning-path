# List 

## Lists in Python are used to store collections of items.  
A list is an ordered, mutable, and iterable data structure that allows duplicate elements. Lists are one of the most commonly used built-in types in Python for grouping related values together.

A list can hold items of any type (numbers, strings, objects, even other lists), and different types can coexist in the same list. Because they are mutable, you can change, add, or remove items after creating the list.

---

**Key features of lists:**
- Defined using square brackets `[]`, with elements separated by commas.
- Maintain the order of elements (insertion order is preserved).
- Allow duplicate values.
- Mutable: items can be added, removed, or changed after creation.
- Elements can be accessed by their index (zero-based).

---

**Common operations:**
- Accessing elements by index (first element is index 0).
- Adding elements with methods like append or extend.
- Inserting items at a specific position.
- Removing items with pop, remove, or clear.
- Slicing to get subsets of the list.
- Checking length with len.
- Iterating through elements using loops.

---

**Use cases of lists:**
- Storing sequences of related data (e.g., a list of student names).
- Collecting items dynamically during runtime.
- Iterating over data collections in loops.
- Serving as simple stacks or queues.
- Handling ordered datasets where duplicates are allowed.

---

👉 Lists are versatile and form the backbone of many Python programs, making them essential for data organization and manipulation.

# 📁 Chaging adding , and removing elements

### Modifying elements 

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

motorcycles[0] = 'ducati'
print(motorcycles)
```

### Adding elements to a list

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

motorcycles.append('ducati')
print(motorcycles)
```

```python
motorcycles = []

motorcycles.append('honda')
motorcycles.append('yamaha')
motorcycles.append('suzuki')

print(motorcycles)
```

### Inserting elements into a list

```python

motorcycles = ['honda', 'yamaha', 'suzuki']

motorcycles.insert(0, 'ducati')
print(motorcycles)
```

### Removing elements from a list

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

del motorcycles[0]
print(motorcycles)
```

### Revoming using pop() 

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

popped_motorcycle = motorcycles.pop()
print(motorcycles)
print(popped_motorcycle)
```

```python

motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

last_own = motorcycles.pop()
print(f'The last motorcycle I owned was a {last_own.title()}.')
```

### Pooping items from any position in a list

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)

last_own = motorcycles.pop(0)
print(f'The last motorcycle I owned was a {last_own.title()}.')
```

### Removing items by value

```python

motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)


motorcycles.remove('yamaha')
print(motorcycles)
```

```python
motorcycles = ['honda', 'yamaha', 'suzuki', 'ducati']
print(motorcycles)

too_expensive = 'ducati'
motorcycles.remove(too_expensive)
print(motorcycles)
print(f'\nA {too_expensive.title()} is too expensive')
```
