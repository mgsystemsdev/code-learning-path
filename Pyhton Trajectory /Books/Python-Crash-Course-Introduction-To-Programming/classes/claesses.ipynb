# 📁 Classes

## Classes in Python are used to model real-world objects and concepts.  
A class is a blueprint for creating objects. It defines the structure (attributes) and behavior (methods) that its objects will have. Classes enable **object-oriented programming (OOP)**, which helps organize and structure complex programs.

Objects created from classes are called **instances**, and each instance can have its own data while sharing behavior defined by the class.

---

**Key features of classes:**
- Defined using the `class` keyword followed by the class name.
- Encapsulate data (attributes) and behavior (methods) together.
- Support inheritance, allowing classes to inherit behavior from other classes.
- Support encapsulation, abstraction, and polymorphism (core OOP principles).
- Allow creation of multiple independent objects based on the same blueprint.

---

**Parts of a class:**
- **Class name**: The identifier used to define and refer to the class. By convention, class names use PascalCase (e.g. `Car`, `Person`).
- **Attributes (fields)**: Variables that hold data about the object. Defined inside the class, often initialized in the `__init__` method.
- **Methods**: Functions defined inside a class that describe the behavior of its objects. They operate on the object’s data.
- **`__init__` method (constructor)**: A special method that runs when a new object is created, used to initialize its attributes.
- **`self` parameter**: Refers to the current instance of the class. Used to access attributes and methods from within the class.
- **Instances (objects)**: Concrete realizations created from the class. Each has its own copy of the class’s attributes.

---

**Common operations:**
- Defining a class with `class`.
- Creating new instances (objects) by calling the class like a function.
- Accessing and modifying object attributes using dot notation.
- Calling object methods using dot notation.
- Inheriting from a base class to create specialized subclasses.
- Overriding methods in subclasses for customized behavior.

---

**Use cases of classes:**
- Modeling real-world entities (like users, cars, products).
- Grouping related data and behavior into one structure.
- Organizing large projects into reusable, modular components.
- Implementing object-oriented design patterns.
- Managing complex data structures with multiple behaviors.

---

👉 Classes are the foundation of object-oriented programming in Python, enabling clean, modular, and scalable code design.


# 📁 Creating and using a class

### Creating a dog class

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def sit(self):
        print(f"{self.name} is sitting now")
    def roll_over(self):
        print(f"{self.name} rolled over")

my_dog = Dog('Willie', 6)

print(f"My dog's name is {my_dog.name}")
print(f"my dog is {my_dog.age} years old")
```

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def sit(self):
        print(f"{self.name} is sitting now")
    def roll_over(self):
        print(f"{self.name} rolled over")

my_dog = Dog('Willie', 6)
my_dog.sit()
my_dog.roll_over()

```

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def sit(self):
        print(f"{self.name} is sitting now")
    def roll_over(self):
        print(f"{self.name} rolled over")

my_dog = Dog('Willie', 6)
your_dog = Dog('Lucy', 3)

print(f"My dog's name is {my_dog.name}")
print(f"my dog is {my_dog.age} years old")
my_dog.sit()
print(f"\nYour dog's name is  {your_dog.name}")
print(f"Your dog is {your_dog.age} years old")
your_dog.sit()
```
