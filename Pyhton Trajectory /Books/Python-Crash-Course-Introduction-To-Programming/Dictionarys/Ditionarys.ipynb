# 📁 Dictonaries 



## Dictionary in python are used to store key-value pairs. A dictionary is a collection which is ordered*, changeable and do not allow duplicates.
A dictionary in Python is a built-in data structure that stores data as key-value pairs. Each key in a dictionary is unique, and is used to access its corresponding value. Dictionaries are mutable, meaning you can change, add, or remove key-value pairs after the dictionary is created.

**Key features of dictionaries:**
- Defined using curly braces `{}` with keys and values separated by colons, e.g. `my_dict = {'key1': 'value1', 'key2': 'value2'}`
- Keys must be immutable types (like strings, numbers, or tuples)
- Values can be any type (strings, numbers, lists, other dictionaries, etc.)
- Dictionaries are unordered (prior to Python 3.7), but from Python 3.7+, they maintain insertion order

**Common operations:**
- Access a value: `my_dict['key1']`
- Add or update a key-value pair: `my_dict['key3'] = 'value3'`
- Remove a key-value pair: `del my_dict['key1']`
- Get all keys: `my_dict.keys()`
- Get all values: `my_dict.values()`
- Get all key-value pairs: `my_dict.items()`

Dictionaries are widely used for fast lookups, configuration storage, and representing structured data.

# 📁 Loops and dictionarys

## A dictionary of similar objects  

```python
favorites_languages = {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'ruby',
    'phil': 'python',
}


language = favorites_languages['sarah'].title()
print(f"Sarah's favorite language is {language}.")
```

## Looping thorught  all keys in dictionary 

```python
for name, language in favorites_languages.items():
    print(f"{name.title()}'s favorite language is {language.title()}.")

for name in favorites_languages.keys():
    print(name.title())



friends = ['phil', 'sarah']
for name in favorites_languages.keys():
    print(f"Hi {name.title()}.")

    if name in friends:
        language = favorites_languages[name].title()
        print(f"\t{name.title()}, I see you love  {language}.")



if 'erin' not in favorites_languages.keys():
    print("Erin, please take our poll!")
```

## Looping through dictionary in particular oder

```python
for name in sorted(favorites_languages.keys()):
    print(f"{name.title()}, thank you for taking the poll.")
```

## Looping through all values in a dictionary

```python
favorites_languages = {
    'jen': ['python', 'ruby'],
    'sarah': ['c'],
    'edward': ['ruby', 'go'],
    'phil': ['python', 'haskell'],
}
 

for name,language in favorites_languages.items():
    print(f"\n{name.title()}'s favorite languages are ")
    for language in language:
        print(f"\t{language.title()}")
```

## A dictionary in a dictionary

```python
users = {
    'aeinstein': {
        'first': 'albert',
        'last': 'einstein',
        'location': 'princeton'
    },

    'mcurie': {
        'first': 'marie',
        'last': 'curie',
        'location': 'paris'
    }
}


for username, user_info  in users.items():
    print(f"\nUsername: {username}")
    full_name = f"{user_info['first']} {user_info['last']}"
    location = user_info['location']
    print(f"\tFull Name: {full_name.title()}")
    print(f"\tLocation: {location.title()}")
```
