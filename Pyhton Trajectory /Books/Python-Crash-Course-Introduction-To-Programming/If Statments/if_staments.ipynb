# 📁 If Staments 



## If statements in Python are used to make decisions in code.

They allow you to execute certain blocks of code only if a condition is true. This is one of the most fundamental control flow tools in Python.

An `if` statement evaluates a condition (an expression that returns `True` or `False`). If the condition is `True`, the indented block of code is executed. If not, the block is skipped.

---

**Key features of `if` statements:**

* Written with the keyword `if`, followed by a condition and a colon.
* Conditions are usually comparisons (e.g. `x > 10`) or Boolean expressions.
* Code inside the `if` block must be indented (usually 4 spaces).
* Can be extended with `elif` (else if) and `else` for multiple decision branches.


---

**Common examples:**

* Simple check:

  ```python
  x = 5
  if x > 0:
    
**Common use cases:**

* Validating inputs
* Handling different cases in logic
* Controlling program flow
* Guard clauses to prevent errors (e.g. checking for `None` or division by zero)

---

👉 `if` statements are the backbone of decision-making in Python, making programs dynamic and responsive.




# 📁 The if 


````python
cars = ['audi', 'bmw', 'subaru', 'toyota']

for car in cars:
    if car == 'bmw':
        print(car.upper())
    else:
        print(car.title())
````

# 📁 The if -elif-else chain

````python

age = 3

if age < 4:
    print("Your admission cost is $0.")
elif age < 18:
    print("Your admission cost is $25.")
else:
    print("Your admission cost is $40.")


age = 12
if age < 4:
    price = 0
elif age < 18:
    price = 25
else:
    price = 40

print(f"Your admission cost is ${price}.")
````

# 📁 Using multiple elif blocks


````python
age = 12
if age < 4:
    price = 0
elif age < 18:
    price = 25
elif age < 65:
    price = 40
else:
    price = 20

print(f"Your admission cost is ${price}.")
````

# 📁 If multiples conditions


### Testing multiple conditions 

````python
requested_toppings = ['mushrooms', 'extra cheese']

if 'mushrooms' in requested_toppings:
    print("Adding mushrooms.")
if 'pepperoni' in requested_toppings:
    print("Adding pepperoni.")
if 'extra cheese' in requested_toppings:
    print("Adding extra cheese.")

print("\nFinished making your pizza!")
````

 ### Using  if statement with lists

````python

requested_toppings = ['mushrooms', 'green peppers', 'extra cheese']
for requested_toppings in requested_toppings:
        print("\n",f"Adding {requested_toppings}.")
print("\nFinished making your pizza!")



requested_toppings = ['mushrooms', 'green peppers', 'extra cheese']
for requested_topping in requested_toppings:
        if requested_topping == 'green peppers':
              print("\n","Sorry, we are out of green peppers right now.")
        else:
              print("\n",f"Adding {requested_topping}.")
print("\nFinished making your pizza!")
````

### Checking that a list not empty

````python
requested_toppings = []

if requested_toppings:
      for requested_topping in requested_toppings:
            print(f"Adding {requested_topping}.")
      print("\nFinished making your pizza!")
else:
      print("Are you sure you want a plain pizza?")
````

### Using multiple lists

````python
available_toppings = ['mushrooms', 'olives', 'green peppers', 'pepperoni', 'pineapple', 'extra cheese']
requested_toppings = ['mushrooms', 'french fries', 'extra cheese']

for requested_topping in requested_toppings:
    if requested_topping in available_toppings:
        print(f"Adding {requested_topping}.")
    else:
        print(f"Sorry, we don't have {requested_topping}.")
print("\nFinished making your pizza!")
````
