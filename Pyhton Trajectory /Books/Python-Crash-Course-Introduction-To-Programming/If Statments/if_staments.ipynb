# If Staments 



## If statements in Python are used to make decisions in code.

They allow you to execute certain blocks of code only if a condition is true. This is one of the most fundamental control flow tools in Python.

An `if` statement evaluates a condition (an expression that returns `True` or `False`). If the condition is `True`, the indented block of code is executed. If not, the block is skipped.

---

**Key features of `if` statements:**

* Written with the keyword `if`, followed by a condition and a colon.
* Conditions are usually comparisons (e.g. `x > 10`) or Boolean expressions.
* Code inside the `if` block must be indented (usually 4 spaces).
* Can be extended with `elif` (else if) and `else` for multiple decision branches.

**Common use cases:**

* Validating inputs
* Handling different cases in logic
* Controlling program flow
* Guard clauses to prevent errors (e.g. checking for `None` or division by zero)

👉 `if` statements are the backbone of decision-making in Python, making programs dynamic and responsive.
